```python
%matplotlib inline
import pandas as pd
import numpy as np
import matplotlib
# matplotlib.use('agg')
import matplotlib.pyplot as plt
import os
# from tqdm.notebook import  tqdm
from tqdm import  tqdm
import talib
import datetime
import math
import  mplfinance as mpf

import sys
sys.path.append('../../DataSource/baostock')
import datasource 
sys.path.append('../..')
import Utils
# 这个是筛选多少天上涨多少的，
codes = datasource.getAStockCodes()
print(f'股票数量:{len(codes)},第一支股票:{codes[0]}')
code = codes[0]
dt = datasource.getData(code)
dt.head()
```

```python
dt['upRate']=(dt['close']-dt['preclose'])/dt['preclose']*100 # 计算涨幅的
for i in [1,3,5]:
    # 计算未来多天的涨幅
    dt[f'next{i}']=(dt['close'].shift(-i)-dt['close'])/dt['close']*100
# 筛选涨停的
dt2 = dt.loc[dt['upRate']>9.5,:] 
for i in [1,3,5]:
    print(f'{i}平均涨幅:{dt[f'next{i}'].mean()},标准差:{dt[f'next{i}'].std()}')
```

```python
# 进行全部统计
dts = []
for i in tqdm(range(len(codes))):
    code = codes[i]
    dt = datasource.getData(code).reindex()
    dt['upRate']=(dt['close']-dt['preclose'])/dt['preclose']*100 # 计算涨幅的
    for j in [1,3,5]:
        # 计算未来多天的涨幅
        dt[f'next{j}']=(dt['close'].shift(-j)-dt['close'])/dt['close']*100
    # 筛选涨停的
    dt2 = dt.loc[dt['upRate']>9.5,:] 
    if len(dt2) > 0:
        dts.append(dt2)
dt_all = pd.concat(dts)
print('完成拼接')
```

```python
dts = []
for i in tqdm(range(len(codes))):
    code = codes[i]
    dt = datasource.getData(code).reindex()
    dt['upRate']=(dt['close']-dt['preclose'])/dt['preclose']*100 # 计算涨幅的
    for j in [1,3,5]:
        # 计算未来多天的涨幅
        dt[f'next{j}']=(dt['close'].shift(-j)-dt['close'])/dt['close']*100
    # 筛选涨停的
    dt2 = dt.loc[dt['upRate']>9.5,:] 
    # 我这里看看每支股票涨停后的情况
    lst = [] # 保存这些数据
    lst.append(code)
    for j in [1,3,5]:
        lst.append(dt2[f'next{j}'].mean())
        lst.append(dt2[f'next{j}'].std())
    lst.append(len(dt2))
    if len(dt2) >0:
        dts.append(lst)
# 这里组成一个数据
dt_all = pd.DataFrame({
    'code':[i[0] for i in dts],
    'next1':[i[1] for i in dts],
    'std1':[i[2] for i in dts],
    'next3':[i[3] for i in dts],
    'std3':[i[4] for i in dts],
    'next5':[i[5] for i in dts],
    'std5':[i[6] for i in dts],
    'count':[i[7] for i in dts]
})
dt_all.to_csv('涨停变化范围.csv')    
```

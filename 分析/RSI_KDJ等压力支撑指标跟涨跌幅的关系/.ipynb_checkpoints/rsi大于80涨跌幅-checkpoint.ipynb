```python
%matplotlib inline
import pandas as pd
import numpy as np
import matplotlib
# matplotlib.use('agg')
# matplotlib.use('TkAgg')

import matplotlib.pyplot as plt
import os
# from tqdm.notebook import  tqdm
from tqdm import  tqdm
import talib
import datetime

import  mplfinance as mpf

import sys
sys.path.append('../../DataSource/baostock')
import datasource
```

```python
# 这个是看rsi大于80的情况下，涨跌幅的
codes = datasource.get_codes()
code = codes[0]
dt = datasource.get_data(code)
dt.head()
```

```python
dt['rsi'] = talib.RSI(dt['close'], timeperiod=14)
dt.head()
```

```python
# 这里要看看涨跌幅的
RSI_LIMIT = 80
DAYS = 5 # 持有多少天
for i in range(DAYS):
    day = i + 1
    # 收盘价滑动
    dt[f'close{day}'] = dt['close'].shift(-day)
    # 计算涨跌幅
    dt[f'rate{day}'] = (dt[f'close{day}'] - dt['close'])/dt['close'] * 100
# 这里需要删除列
dt = dt.drop([f'close{i+1}' for i in range(day)], axis=1)
dt.head()
```

```python
dt2 = dt.loc[dt['rsi'] > RSI_LIMIT]
dt2.head()
```

```python
dt2.loc[:,[f'rate{i+1}' for i in range(DAYS)]].sum()
```

```python
dt2.loc[:,[f'rate{i+1}' for i in range(DAYS)]].std()
```

```python
# 我这里需要看看范围具体是怎样的
COLUMNS = ['rsi']
COLUMNS.extend(f'rate{i+1}' for i in range(DAYS))
dt3 = dt2.loc[:,COLUMNS].astype(int)
dt3.head()
```

```python
# 我看看比例分配
dt4 = dt3.groupby(f'rate1').size().reset_index().rename(columns={0:'count'}).sort_values(by='rate1', ascending=True)
dt4.head()
```

```python
# 然后这里画柱状图吧
plt.bar(dt4['rate1'], dt4['count'])
```

```python
plt.figure(figsize=(15, 12))
dt4 = dt3.groupby(f'rate1').size().reset_index().rename(columns={0:'count'}).sort_values(by='rate1', ascending=True)
plt.subplot(2, 2, 1)
plt.bar(dt4['rate1'], dt4['count'])
plt.title("1")

dt4 = dt3.groupby(f'rate2').size().reset_index().rename(columns={0:'count'}).sort_values(by='rate2', ascending=True)
plt.subplot(2, 2, 2)
plt.bar(dt4['rate2'], dt4['count'])
plt.title("2")

dt4 = dt3.groupby(f'rate3').size().reset_index().rename(columns={0:'count'}).sort_values(by='rate3', ascending=True)
plt.subplot(2, 2, 3)
plt.bar(dt4['rate3'], dt4['count'])
plt.title("3")

dt4 = dt3.groupby(f'rate4').size().reset_index().rename(columns={0:'count'}).sort_values(by='rate4', ascending=True)
plt.subplot(2, 2, 4)
plt.bar(dt4['rate4'], dt4['count'])
plt.title("4")
```

# 这里开始总结

```python

# 这里要看看涨跌幅的
RSI_LIMIT = 95
DAYS = 5 # 持有多少天
COLUMNS = ['rsi']
COLUMNS.extend(f'rate{i+1}' for i in range(DAYS))
lst_dt_rsi = [[] for i in range(DAYS)] # 多个
for i in tqdm(range(len(codes))):
    code = codes[i]
    dt = datasource.get_data(code)
    # 这里先计算指标
    dt['rsi'] = talib.RSI(dt['close'], timeperiod=14)
    # 然后涨跌幅
    for i in range(DAYS):
        day = i + 1
        # 收盘价滑动
        dt[f'close{day}'] = dt['close'].shift(-day)
        # 计算涨跌幅
        dt[f'rate{day}'] = (dt[f'close{day}'] - dt['close'])/dt['close'] * 100
    # 筛选
    dt2 = dt.loc[dt['rsi'] > RSI_LIMIT,COLUMNS].dropna()
    # 转换成整数
    dt3 = dt2.astype(int)
    for i in range(DAYS):
        dt4 = dt3.groupby(f'rate{i+1}').size().reset_index().rename(columns={0:'count'}).sort_values(by=f'rate{i+1}', ascending=True)
        lst_dt_rsi[i].append(dt4)
```

```python
# 这里需要先汇总
import math
dt_alls =[ pd.concat(dts,  ignore_index=True) for dts in lst_dt_rsi]
# 图的大小
plt.figure(figsize=(15, 12))
img_column = 2 # 每行2列
img_row = math.ceil(len(dt_alls) / img_column)

# 绘制子图
for i in range(len(dt_alls)):
    # 这里需要再次计算并重新排序的，请注意这里是sum
    dt4 = dt_alls[i].groupby(f'rate{i+1}').sum().reset_index().rename(columns={0:'count'}).sort_values(by=f'rate{i+1}', ascending=True)
    plt.subplot(img_row, img_column, i+1)
    plt.bar(dt4[f'rate{i+1}'], dt4['count'])
    plt.title(f"{i+1}")
    # 这里输出均值,这里用加权算术平均
    dt4['sum1'] = dt4[f'rate{i+1}'] * dt4['count'] # 
    _sum1 = dt4['sum1'].sum()
    _sum2 = dt4['count'].sum()
    print(f'{i+1}加权算术均值:{_sum1/_sum2}')
```

# 看看大于某个阈值的图形到底是什么样子

```python
dest_dir = './rsi大于阈值图形'
# 这里要看看涨跌幅的
RSI_LIMIT = 95
DAYS = 5 # 持有多少天
COLUMNS = ['rsi']
COLUMNS.extend(f'rate{i+1}' for i in range(DAYS))
day_before = 30
days_after = 7
lst_dt_rsi = [[] for i in range(DAYS)] # 多个
for i in tqdm(range(len(codes))):
    code = codes[i]
    dt = datasource.get_data(code)
    # 这里先计算指标
    dt['rsi'] = talib.RSI(dt['close'], timeperiod=14)
    dt = dt.dropna()
    # 这里遍历一下大于某个值的
    for index in dt2.index:
        # 这里要判断是否大于某个值
        if dt.at[index,'rsi'] >= RSI_LIMIT:
            _before = index - datetime.timedelta(days=day_before) # 前面几天
            _after = index + datetime.timedelta(days=days_after)   # 后边几天
            _df_plot = dt.loc[_before:_after,:]                      # 截取
            mydpi = 96
            fig1, _ = mpf.plot(_df_plot,
                           type='candle',
                           style='yahoo',
                           figsize =(800/mydpi,600/mydpi), 
                           volume=True,
                           axisoff=True,
                           returnfig=True, 
                           scale_padding=0.2)
            # 然后这里是保存
            _img_file = f'{code}-{index.strftime("%Y-%m-%d")}.jpg'
            _img_path = os.path.join(dest_dir, _img_file)
            fig1.savefig(_img_path,dpi=mydpi)
            plt.clf()
            plt.close()
```

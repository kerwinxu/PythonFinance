我想要看看均线中上涨或者下跌的统计

```python
%matplotlib inline
import pandas as pd
import numpy as np
import matplotlib
# matplotlib.use('agg')
import matplotlib.pyplot as plt
import os
# from tqdm.notebook import  tqdm
from tqdm import  tqdm
import talib
import datetime
import math
import  mplfinance as mpf

import sys
sys.path.append('../../DataSource/聚宽数据')
import datasource 
sys.path.append('../..')
import Utils
# 这个是筛选多少天上涨多少的，
codes = datasource.getAStockCodes()
print(f'股票数量:{len(codes)},第一支股票:{codes[0]}')
code = codes[0]
dt = datasource.getData(code)
dt.head()
```

```python
N_DAYS = 10 # 均线
dt.loc[:,f'EMA{N_DAYS}'] = talib.EMA(dt['close'], timeperiod=N_DAYS)
# 我我想看如果今天的均线上升，那么收盘价是什么情况
dt.loc[:, 'pre_close'] = dt['close'].shift(1)
dt.loc[:, f'pre_EMA{N_DAYS}']=dt.loc[:,f'EMA{N_DAYS}'].shift(1)
# 我只是想看平均值，中位值
dt.loc[:, 'rate_close_pre'] = (dt['close'] - dt['pre_close']) / dt['pre_close'] * 100
dt2 = dt.loc[dt[f'EMA{N_DAYS}'] > dt[ f'pre_EMA{N_DAYS}'] , :]
print(f'均线上涨时收盘价涨幅均值:{dt2['rate_close_pre'].mean()}, 中位值:{dt2['rate_close_pre'].median()}')
dt3 = dt.loc[dt[f'EMA{N_DAYS}'] < dt[ f'pre_EMA{N_DAYS}'] , :]
print(f'均线下跌时收盘价涨幅均值:{dt3['rate_close_pre'].mean()}, 中位值:{dt3['rate_close_pre'].median()}')
```

```python
# 这里看一下多种均线的情况
for i in range(30):
    N_DAYS = 2 + i # 这个作为均线
    dt.loc[:,f'EMA{N_DAYS}'] = talib.EMA(dt['close'], timeperiod=N_DAYS)
    # 我我想看如果今天的均线上升，那么收盘价是什么情况
    dt.loc[:, 'pre_close'] = dt['close'].shift(1)
    dt.loc[:, f'pre_EMA{N_DAYS}']=dt.loc[:,f'EMA{N_DAYS}'].shift(1)
    # 我只是想看平均值，中位值
    dt.loc[:, 'rate_close_pre'] = (dt['close'] - dt['pre_close']) / dt['pre_close'] * 100
    dt2 = dt.loc[dt[f'EMA{N_DAYS}'] > dt[ f'pre_EMA{N_DAYS}'] , :]
    dt3 = dt.loc[dt[f'EMA{N_DAYS}'] < dt[ f'pre_EMA{N_DAYS}'] , :]
    print(f"{N_DAYS:<2}均线上涨时收盘价涨幅均值:{dt2['rate_close_pre'].mean():.2f}, 中位值:{dt2['rate_close_pre'].median():.2f}，均线下跌时收盘价涨幅均值:{dt3['rate_close_pre'].mean():.2f}, 中位值:{dt3['rate_close_pre'].median():.2f}")
```
